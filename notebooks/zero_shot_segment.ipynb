## Visualize token decomposition on full output

```python
## Get the texts
lines = ["alphabet letter", "dog", "cat"]
texts = tokenizer(lines).to(device)  # tokenize text for encoder
class_embeddings = model.encode_text(texts, normalize=True)  # get text embeddings
```

```python
def plot_segment_map(attentions, mlps, class_embeddings, image_pil, lines, device, layer=None, head=None, data=None):
    """
    Generate and display zero-shot segmentation maps for an arbitrary number of text embeddings.
    
    Parameters:
        attentions (torch.Tensor): Attention maps from the model with shape [B, heads, tokens, tokens].
                                   (Assumed that the first token is the CLS token.)
        class_embeddings (torch.Tensor): Text embeddings for each class with shape [num_classes, embed_dim].
        image_pil (PIL.Image): The original image (as a PIL image) on which to overlay the segmentation maps.
        lines (list of str): Class names corresponding to the class embeddings.
        device: The torch device to perform interpolation on.
        layer (int, optional): If specified, use the attention from this layer index.
        head (int, optional): If specified (and layer is also specified), use the attention from this head index.
        data (list, optional): If provided, the contribution look at ar the one in the heads of data
    """
    # ------------------------------
    # 1. Compose the attention map
    # ------------------------------

    if layer is None:
        pred_full = attentions[0, :, :, 1:].sum(1) @ class_embeddings.T
        """ # Sum over all layers and heads, skip the CLS token.
        pred = attentions[0, :, :, 1:].sum(axis=(0, 1))
        attention_map = pred / pred.norm(dim=-1, keepdim=True) @ class_embeddings.T """

        attention_map, idx = torch.max(pred_full, dim=0)
        attention_map /= attention_map.norm(dim=0, keepdim=True)
        
        print(attention_map.shape)
    else:
        # Use the specified layer
        if head is None:
            # Sum over all heads in that layer, skip the CLS token
            pred = attentions[0, layer, :, 1:].sum(axis=0) 
            attention_map = pred / pred.norm(dim=-1, keepdim=True) @ class_embeddings.T
        else:
            # Use the specified head in that layer
            pred = attentions[0, layer, head, 1:]
            attention_map = pred / pred.norm(dim=-1, keepdim=True) @ class_embeddings.T
    if data:
        pred = None
        for c, entry in enumerate(data):
            layer = entry["layer"]
            head = entry["head"]
            if c == 0:
                pred = attentions[0, layer, head, 1:]
            else:
                pred += attentions[0, layer, head, 1:]

    # ------------------------------
    # 2. Recover spatial dimensions
    # ------------------------------
    num_patches = attention_map.shape[0]
    dim_patch = int(np.sqrt(num_patches))  # assume a square grid
    # ------------------------------
    # 3. Rearrange and upscale
    # ------------------------------

    attention_map = einops.rearrange(
        attention_map,
        '(N M) C -> 1 C N M',
        N=dim_patch, M=dim_patch
    )
    
    width, height = image_pil.size  # PIL returns (width, height)
    
    attention_map = F.interpolate(
        attention_map,
        size=(height, width),
        mode='bilinear'
    ).to(device)
    
    # ------------------------------
    # 4. Convert to NumPy
    # ------------------------------
    # Shape: [1, num_classes, H, W] -> [num_classes, H, W]
    attention_map = attention_map[0].detach().cpu().numpy()
    num_classes = attention_map.shape[0]
    
    print("Attention map shape (num_classes, H, W):", attention_map.shape)

    # -----------------------------------------------------------------
    # (A) Per-class segmentation overlays
    # -----------------------------------------------------------------
    for i in range(num_classes):
        diff = attention_map[i]
        # Normalize each class overlay for display
        diff = (diff - diff.min()) / (diff.max() - diff.min() + 1e-8)
        diff = diff > 0.5
        plt.figure()
        plt.imshow(image_pil)
        plt.imshow(diff, cmap='jet', alpha=0.5, vmin=0, vmax=1)
        plt.title(f"Class {lines[i]} Segmentation")
        plt.colorbar()
        plt.axis('off')
        plt.show()

    # -----------------------------------------------------------------
    # (B) Final "hard" segmentation visualization
    # -----------------------------------------------------------------
    # Normalize each with each other
    # 5. Get the top-scoring class for each pixel
    predicted_classes = np.argmax(attention_map, axis=0)  # shape: (H, W)

    # 6. Create (or retrieve) a color palette
    cmap = plt.get_cmap("tab20", num_classes)
    
    # 7. Create a colored mask
    colored_mask = np.zeros((height, width, 3), dtype=np.uint8)
    for c in range(num_classes):
        # Each channel is in [0, 1], so convert to [0, 255]
        r, g, b, _ = [int(255 * x) for x in cmap(c)]
        colored_mask[predicted_classes == c] = [r, g, b]

    # 8. Blend the mask with the original image
    alpha = 0.3
    image_np = np.array(image_pil)
    if image_np.ndim == 2:  # If grayscale, expand to 3 channels
        image_np = np.stack([image_np]*3, axis=-1)

    blended = (alpha * image_np.astype(float) + (1 - alpha) * colored_mask.astype(float)).astype(np.uint8)

    # ------------------------------
    # 9. Show final segmentation + legend
    # ------------------------------
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.imshow(blended)
    ax.set_title("Hard Segmentation Visualization (Top Class per Pixel)")
    ax.axis('off')

    # Create a patch for each class to display in legend
    legend_patches = []
    for c in range(num_classes):
        # For the legend, we can directly use `cmap(c)`, which is RGBA in [0..1]
        color = cmap(c)
        patch = mpatches.Patch(color=color, label=lines[c])
        legend_patches.append(patch)
    
    # Place legend outside the image (e.g., top-right)
    ax.legend(
        handles=legend_patches, 
        bbox_to_anchor=(1.05, 1), 
        loc='upper left', 
        borderaxespad=0.
    )

    plt.show()

    # Print final prediction
    prediction = ((attentions.sum(axis=(1, 2, 3)) + mlps.sum(axis=1))@ class_embeddings.T)
    print(prediction.shape)

    print(f"Final prediction is {lines[prediction.argmax(-1)]}. General logits are {prediction} for labels {lines}")
```

```python
# Plot last layers
num_last_layers_plot = 2
for k in range(nr_layers_ - num_last_layers_plot, nr_layers_):
    print(f"Layer {k}")
    plot_segment_map(attentions, mlps, class_embeddings, image_pil, lines, device, layer=k)
print("Final sum of all")
plot_segment_map(attentions, mlps,  class_embeddings, image_pil, lines, device)
```
